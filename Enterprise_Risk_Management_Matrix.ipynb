## 10. Validation Summary

Final validation checks to ensure the model meets requirements.

```python
from IPython.display import HTML, display

# Validation checks
validation_results = []

# Check 1: Risk decomposition hierarchy exists
enterprise_risks = [n for n in graph.nodes if n.node_type == 'Enterprise Risk']
validation_results.append(('Risk Hierarchy', len(enterprise_risks) > 0, f"{len(enterprise_risks)} enterprise risk node(s)"))

# Check 2: All risks have formal statements
all_have_statements = all(hasattr(r, 'cause') and hasattr(r, 'risk_event') and hasattr(r, 'business_impact') for r in model.risks)
validation_results.append(('Formal Risk Statements', all_have_statements, f"{len(model.risks)} risks with complete statements"))

# Check 3: All risks have controls
risks_with_controls = len(set(risk_id for control in model.controls for risk_id in control.risks_mitigated))
validation_results.append(('Risk Coverage', risks_with_controls == len(model.risks), f"{risks_with_controls}/{len(model.risks)} risks have controls"))

# Check 4: Platform implementations exist
platforms_covered = len(set(i.platform for i in model.implementations))
validation_results.append(('Platform Coverage', platforms_covered >= 3, f"{platforms_covered} platforms"))

# Check 5: Evidence artifacts exist
validation_results.append(('Evidence Artifacts', len(model.evidence_artifacts) > 0, f"{len(model.evidence_artifacts)} evidence items"))

# Check 6: Regulatory mappings exist
regulations_covered = len(set(m.regulation for m in model.regulatory_mappings))
validation_results.append(('Regulatory Coverage', regulations_covered >= 5, f"{regulations_covered} regulations mapped"))

# Check 7: Graph is connected
validation_results.append(('Graph Connectivity', len(graph.edges) > 0, f"{len(graph.edges)} relationships"))

# Display results
html_output = "<h3>Validation Summary</h3><table style='border-collapse: collapse; width: 100%;'>"
html_output += "<tr style='background-color: #f0f0f0;'><th style='padding: 10px; border: 1px solid #ddd;'>Check</th><th style='padding: 10px; border: 1px solid #ddd;'>Status</th><th style='padding: 10px; border: 1px solid #ddd;'>Details</th></tr>"

for check_name, passed, details in validation_results:
    status_icon = "✓" if passed else "✗"
    status_color = "green" if passed else "red"
    html_output += f"<tr><td style='padding: 10px; border: 1px solid #ddd;'>{check_name}</td>"
    html_output += f"<td style='padding: 10px; border: 1px solid #ddd; color: {status_color}; font-weight: bold;'>{status_icon}</td>"
    html_output += f"<td style='padding: 10px; border: 1px solid #ddd;'>{details}</td></tr>"

html_output += "</table>"

display(HTML(html_output))

# Overall status
all_passed = all(result[1] for result in validation_results)
if all_passed:
    display(HTML("<div style='background-color: #d4edda; border: 1px solid #c3e6cb; padding: 15px; margin-top: 20px; border-radius: 5px;'><h4 style='color: #155724; margin: 0;'>✓ All validation checks passed</h4><p style='color: #155724; margin: 5px 0 0 0;'>The ERM system meets all requirements and is ready for use.</p></div>"))
else:
    display(HTML("<div style='background-color: #f8d7da; border: 1px solid #f5c6cb; padding: 15px; margin-top: 20px; border-radius: 5px;'><h4 style='color: #721c24; margin: 0;'>✗ Some validation checks failed</h4><p style='color: #721c24; margin: 5px 0 0 0;'>Please review and address the failures above.</p></div>"))
```
